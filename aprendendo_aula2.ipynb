```python
import numpy as np
import random
np.random.seed(42)
array=np.random.randint(0,100,10)
print('Array: ',array)
print('Maior valor do array: ',array.max())
print('Posição do maior valor do array: ',array.argmax())
print('Menor valor do array: ',array.min())
print('Posição do menor valor do array: ',array.argmin())
print('Soma dos elementos: ',array.sum())
print('Média dos valores: ',array.mean())
print('Desvio padrão dos elementos: ',array.std())
```

```python
indices=np.argsort(array)
print('Array: ',array)
print('Indices: ',indices)
```

```python
array.sort()
print('Array ordenado: ',array[::-1])
print('Array ordenado: ',array[0::])
print('Muda para string: ',array.astype(str))
```

```python
array=np.arange(18).reshape((3,6))
print('Matriz:')
print(array)
```

```python
mascara = array>7
mascara
```

```python
array[mascara]
```

```python
mask = (array>7)
print('Máscara booleana da matriz para valores maiores que 7')
print(mask)
print('\nRecuperando apenas os elementos maiores que 7:')
print(array[mask])
print('\nModificando os elementos maiores que 7:')
array[mask]=0
print(array)
```

```python
import numpy as np
import random
np.random.seed(42)
aleat_int=np.random.randint(0,100,20)
print(aleat_int)
```

```python
pares=aleat_int%2==0
print(pares)
```

```python
aleat_int[pares]
```

```python
aleat_int[~pares]=0
aleat_int
```

```python
maiores_50=aleat_int>50
print(maiores_50)
```

```python
print(pares&maiores_50)
```

```python
print(aleat_int[pares&maiores_50])
```

```python
print(aleat_int[(aleat_int%2==0)&(aleat_int>50)])
```

```python
print(aleat_int[(aleat_int%2==0)|(aleat_int>50)])
```
