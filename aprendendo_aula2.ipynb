Exercício 4)
Construa um array A bidimensional 5x6 com 30 números inteiros escolhidos de forma randômica no intervalo entre 0 e 20.

faça uma cópia do array A criado, chamando a cópia de Ac
Substitua todos os elementos do array A que sejam maiores que 10 pelo valor -1
Substitua todos os elementos do array Ac que sejam maiores que 7 e menores que 15 pelo valor -1
Dica: Gere máscaras booleanas para realizar as modificações nos arrays.

```python
import numpy as np
# Criando a matriz A (5x6) com valores aleatórios entre 0 e 20
A = np.random.randint(0, 21, (5, 6))
print("Matriz A original:")
print(A)
# Criando uma cópia de A
Ac = A.copy()
# Substituindo em A os valores maiores que 10 por -1
A[A > 10] = -1
# Substituindo em Ac os valores maiores que 7 e menores que 15 por -1
Ac[(Ac > 7) & (Ac < 15)] = -1
print("\nMatriz A modificada:")
print(A)
print("\nMatriz Ac modificada:")
print(Ac)
```

Exercício 5)
Construa um array A com 20 linhas e 10 colunas onde os elementos são os números interios de 1 até 200.

Crie uma view de A chamada A_lpares contendo apenas as linhas de A com índice par.
Crie uma view de A chamada A_lpares_cimpares contendo as linhas A com índice par e colunas com índice ímpar.

Dica: Construa uma máscara booleana

```python
import numpy as np
# Criando a matriz A (20x10) com valores inteiros aleatórios entre 1 e 200
A = np.random.randint(1, 201, (20, 10))
print("Matriz A original:")
print(A)

# Máscara para linhas com índice par
linhas_pares = np.arange(A.shape[0]) % 2 == 0
# View com linhas de índice par
A_lpares = A[linhas_pares, :]
print("\nA_lpares:")
print(A_lpares)
# Máscara para colunas com índice ímpar
colunas_impares = np.arange(A.shape[1]) % 2 == 1
# View com linhas pares e colunas ímpares
A_lpares_cimpares = A[linhas_pares][:, colunas_impares]
print("\nA_lpares_cimpares:")
print(A_lpares_cimpares)
```

Exercício 6)
Substitua os elememtos do array A do exercício anterior por -1 quando pelo menos um dos índices do elemento é par.

Dica: Construa uma máscara booleana

```python
import numpy as np
# Matriz A com números de 1 a 200
A = np.arange(1, 201).reshape(20, 10)
# Índices das linhas e colunas
linhas = np.arange(A.shape[0]).reshape(-1, 1)
colunas = np.arange(A.shape[1]).reshape(1, -1)
# Máscara: linha par OU coluna par
mascara = (linhas % 2 == 0) | (colunas % 2 == 0)
# Substitui por -1 onde a máscara é True
A[mascara] = -1
print(A)
```

Exercício 7)
Construa uma função chamada ordena_linhas que recebe uma matriz como parâmetro e troque as linhas de posição para que fiquem ordenadas de acordo com a primeira coluna. Por exemplo:

A = np.array([[20, 26, 12],
              [8, 3, 26],
              [3, 10, 4]])
print(ordena_linhas(A))
deve resultar em

[[ 3 10  4]
 [ 8  3 26]
 [20 26 12]]

```python
import numpy as np
def ordena_linhas(A):
  return A[np.argsort(A[:, 0])]
# Exemplo
A = np.array([[20, 26, 12],
[8, 3, 26],
[3, 10, 4]])
print(ordena_linhas(A))
```
